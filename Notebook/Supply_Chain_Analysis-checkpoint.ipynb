```python
# identify inventory risks
#How many products are out of stock?
#How many products have critically low stock (10 or below)?
#Which product categories have the highest inventory risk?
inventory_analysis = df.groupby("Product type").agg(
    Total_Products=("SKU", "count"),
    Average_Stock=("Stock levels", "mean"),
    Low_Stock_Products=("Stock levels", lambda x: (x <= 10).sum()),
    Out_of_Stock=("Stock levels", lambda x: (x == 0).sum()),
    Average_Units_Sold=("Number of products sold", "mean")
).round(2)

inventory_analysis
```

```python
#Identify high-demand products with low stock
inventory_risk = df[
    (df["Stock levels"] <= 10) &
    (df["Number of products sold"] > df["Number of products sold"].mean())
]

inventory_risk[
    ["SKU", "Product type", "Stock levels", "Number of products sold"]
].sort_values(
    by="Number of products sold",
    ascending=False
)
```

```python
# : Revenue distribution analysis
revenue_analysis = df.groupby("Product type").agg(
    Total_Revenue=("Revenue generated", "sum"),
    Average_Revenue=("Revenue generated", "mean"),
    Minimum_Revenue=("Revenue generated", "min"),
    Maximum_Revenue=("Revenue generated", "max")
).round(2)

revenue_analysis.sort_values(
    by="Total_Revenue",
    ascending=False
)
```

```python
# Cost analysis
# Let's compare manufacturing costs, shipping costs and overall recorded costs across product categories.

cost_analysis = df.groupby("Product type").agg(
    Average_Manufacturing_Cost=("Manufacturing costs", "mean"),
    Average_Shipping_Cost=("Shipping costs", "mean"),
    Average_Total_Cost=("Costs", "mean"),
    Total_Recorded_Cost=("Costs", "sum")
).round(2)

cost_analysis
```

```python
# Outlier detection
# We'll identify unusually high or low values in important numerical columns using the IQR method.
columns = [
    "Price",
    "Number of products sold",
    "Revenue generated",
    "Stock levels",
    "Manufacturing costs",
    "Shipping costs",
    "Defect rates",
    "Costs"
]

outlier_summary = []

for col in columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    outlier_summary.append({
        "Column": col,
        "Outlier_Count": len(outliers),
        "Lower_Limit": round(lower_limit, 2),
        "Upper_Limit": round(upper_limit, 2)
    })

outlier_df = pd.DataFrame(outlier_summary)
outlier_df
```
